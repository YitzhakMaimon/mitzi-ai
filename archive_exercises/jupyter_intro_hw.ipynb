Use the right arithmetical operator between the variables. Find the index of the space character, you can use .find() method to find it. Use the index of space that you found and print in f string the length of the first name and the length of the last name.

```python
#QUESTION8

a = '#####'
b = '*****'
c = 'Yitzhak Maimon'
result = a + c + b
index_space = result.find(" ")
length_first_name = index_space - 5
length_last_name = len(result) - index_space -5 - 1
print(f"the length of first name is {length_first_name} and length of last name is {length_last_name}")
```
